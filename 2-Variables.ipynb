## Variables In Python 

Variables are fundamental elements in programming used to store data that can be referrenced and manipulated in a program.
In python, variables are created when we assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when we assign a value to the variable.


```python
# Declaring and assinging variables

integer = 10
decimal = 10.2
string = "abhi"
boolean = True
```

```python
# Naming Conventions

# variable name should be descriptive.
# they must start with letter or "_" and can contain letters, numbers and uderscore.
# variable names are case sensitive 

# EXAMPLES - Correct naming conventions 
name = "abhi"
_name = "abhi"

print(name)  


# EXAMPLES - wrong naming conventions 
  # 1name = "abhi"
  # @name = "abhi"
Name = "abhi"
print(Name) # correct 
# print(NAME) wrong because "NAME" and "Name" are considered as two different variables

```

```python
# Understanding variable type
# Python is dynamically typed : type of variable is determined at runtime 

print(integer, type(integer))
print(decimal, type(decimal))
print(string, type(string))
print(boolean, type(boolean))

# type checking is done using a in-built method : type()
```

```python
# Type conversion 

integer = 10
print(integer, " & type of integer : ", type(integer))

string = str(integer)
print(string, " & type of string : ", type(string))

decimal = float(string)
print(decimal, " & type of decimal : ", type(decimal))

# name = "abhi" 
# number = int(name)
# this type conversion fails because for coversion there must be a string of numerical values
```

```python
# Dynamic Typing 
# Python allows to change the type of variable as the program executes

var = 20
print(var, type(var))

var = "abhi"
print(var, type(var))

var = 10.5
print(var, type(var))

var = True
print(var, type(var))
```
