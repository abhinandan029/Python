# Sets

Sets are built-in data types in python used to store collection of uique items. They are unorderd, means that the elements do not follow a specific order, and they do not allow duplicate elements. 



```python
# creating a set 

my_set = set([1, 2, 3])
print(my_set)

newSet = set([1, 2, 3, 5, 5, 6, 6])
print(newSet)
```

## Operations on Sets 


```python
# 1. add(ele) : addig a element to the set

newSet.add(4)
print(newSet)

# 2. remove(ele) : removes the element and gives an error when ele is not present 
newSet = { 1, 2, 3, 4, 5, 6 ,7} 
newSet.remove(1) 
print(newSet)

# 3. discard(ele) : removes the element but does not provide an error when ele is not present 
newSet.discard(1)
newSet.discard(4)
print(newSet)

# 4. pop() : Removes the first element and returns it
newSet = {1, 2, 3, 4, 5, 6, 7}
ele = newSet.pop()
print(ele, newSet)

# 5. clear() : removes all the elements from set
newset = {1, 2, 3}
newSet.clear()
print(newSet)
```

```python
# use of 'in' key word : returns true or false

newSet = {1, 2, 3, 4, 5}
print(1 in newSet)
print(6 in newSet)
```

## Set Operations 

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# 1. Union : combines both the sets 
union = set1.union(set2)
print(union)

# 2. Intersection : returns the common elements between both sets
intersection = set1.intersection(set2)
print(intersection)

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# 3. Intersection_update : finds the common elements add it to first the set
set1.intersection_update(set2)
print(set1)

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# 4. difference : returns the elements of first set by removing the common elements present in second set
difference = set1.difference(set2)
print(difference)

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# 5. dfference_update
set1.difference_update(set2)
print(set1)

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# 6. symmetric difference : common elements are removed and combines both the sets 
print(set1.symmetric_difference(set2))
```

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}

# issubset()
print(set1.issubset(set2))

# issupperset
print(set1.issuperset(set2))
```
