# Dictionaries 

Dictionaries are unordered collection of items. They store data in key value pairs. Keys must be unique and are immutable, while values can be of any type.

```python
# creating a empty dictionary

dictionary = dict()
print(dictionary)

dictionary = { "name" : "Abhi", "age" : 21}
print(dictionary)

```

```python
# accessing elements of dictionary 

print(dictionary['name'])

# OR 
print(dictionary.get('name'))
```

```python
# updating elements in the dictionary 

dictionary['name'] = "Abhinandan"
dictionary['grade'] = "A"

print(dictionary)
```

```python
# deleting element from the dictionary

del dictionary['grade']
print(dictionary)
```

## Methods on Dictionary

```python
# 1. keys() : returns all the keys of a dictionary
keys = dictionary.keys()
print(keys)

# 2. values() : returns all the values of a dictionary
values = dictionary.values()
print(values)

# 3. items() : return all key value pairs in tuples
items = dictionary.items()
print(items)
```

```python
# shallow copy 

#without shallow copy
copy = dictionary
print(copy)
print(dictionary)

dictionary['name'] = "Abhi"   # without shallow copy it updates the value of 'name'everywhere
print(copy)
print(dictionary)

#with shallow copy 
copy2 = dictionary.copy()
print(copy2)
print(dictionary)

dictionary['name'] = "Abhinandnan"
print(copy2)
print(dictionary)
```

```python
# Nested Dictionary 

students = {
  "student1" : {"name" : "Abhi", "age" : 18},
  "student2" : {"name" : "Ram", "age" : 20}
}

print(students["student1"]["name"], students["student1"]["age"])
print(students["student2"]["name"], students["student2"]["age"])
```

```python
# Dictionary Comprehension

squares = {x : x**2 for  x in range(5)}
print(squares)

even_squares = {x : x**2 for x in range(10) if x % 2 == 0}
print(even_squares)
```
