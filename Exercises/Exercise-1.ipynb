# Simple Calculator

```python
num_1 = float(input("Enter the first number : "))
num_2 = float(input("Enter the second number : "))

operator = input("select he operator ( +, -, *, / )")

if operator == "+" :
  print(num_1 + num_2)

elif operator == "-" : 
  print(num_1 - num_2)

elif operator == "*" :
  print(num_1 * num_2)

elif operator == "/":
  if num_2 == 0 : 
    print("second number cannot be zero")
  else :
    print(num_1 / num_2)
  
else : 
  print("please enter a valid operator.")
  
```
