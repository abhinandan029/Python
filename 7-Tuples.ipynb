# Tuples 

tuples are ordered collection of items that are immutable.

```python
#  creating a tuple 
new_tuple  = ()
print(new_tuple)

# OR
new_tuple_2 = tuple()
print(new_tuple_2) 
```

```python
# converting a list to tuple 
list = [1, 2, 3, 4, 5]

tuple = tuple(list)
print(tuple)
```

```python
# mixed tuple
mixed_tuple = (1, "hello", 1.5)

print(mixed_tuple)

# Indexing in tuples 

print(mixed_tuple[0])
print(mixed_tuple[1])
print(mixed_tuple[2])

# slicing operation are same as list
```

## Operations on Tuples

```python
# 1. Concatination : adds two tuples

tuple_1 = (1, 2, 3)
tuple_2 = (4, 5, 6)

result = tuple_1 + tuple_2

print(result)
```

```python
# 2. Scalar Multiplication : repeats the tuple for the scaler number of tuple 

tuple = (1, 2, 3)

result = 3 * tuple
print(result)
```

## Methods on Tuples

```python
# 1. count(ele) : returns the number of repeatations of an element 
numbers = (1, 2, 2, 3, 3, 3)
print(numbers.count(1), numbers.count(2), numbers.count(3))

# 2. index(ele) : returns the index of first occurence of an element

print(numbers.index(1), numbers.index(2), numbers.index(3))
```

```python
# packing and unpacking the tuples

#packing
tuple = 1, 2, 3
print(tuple)

#unpacking : for this number of vairable must be equal to number of elements in tuple
a, b, c = tuple
print(f"a = {a}, b = {b}, c = {c}")

# unpacking with * : here the number of variable does not matter
tuple = 1, 2, 3, 4, 5
start, *middle, end = tuple
print(start, middle, end)
```

```python
# Nested tuple

tuple = (1, (1, 2), (1, 2, 3))
print(tuple)

print(tuple[0])
print(tuple[1])
print(tuple[2])

# accessing element inside a nested tuple
print(tuple[1][0], tuple[1][1])

# accessing through slicing
print(tuple[2][0:3])
```
