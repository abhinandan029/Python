# Loops

## FOR loop

```python
for i in range(5) :
  print(i)
```

```python
# range(start, stop, step)
# start : the index from where the loop starts.
# stop : the index where the loop stops. 
# step : the amount of value needs to be incrimented per iteration.

for i in range(1, 11, 1) :
  print(i)
```

```python
for i in range(10, 0, -1) :
  print(i)
```

```python
string = "Abhi M"

for i in string :
  print(i)
```

## WHILE loop

```python
# The while loop continues to execute as long as condition is true 

count = 0

while count <= 5 :
  print(count)
  count = count + 1
```

### Loop control statements

```python
# BREAK statement : exits the loop completely when an condition is satisfied.

for i in range(10) :
  
  if i == 5 :
    print("here we wil exit the loop when i = ", i)
    break
  print(i)
```

```python
# CONTINUE statement : skips the current condition and continue with next iteration.

for i in range(11) :
  
  if i % 2 == 0 :
    print("This number is skipped.", i)
    continue
  
  print(i)
```

```python
# PASS statement : it is a  null operation, it does nothing 

for i in range(11) :
  if i == 5 :
    print("In the pass statement.")
    pass
  
  print(i)
```

## NESTED Loops

```python
for i in range(3) :
  for j in range(2) :
    print(f"i : {i} and j : {j}")
```
