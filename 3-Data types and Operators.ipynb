# Data Types

## Definition : 
- Data types are the classification of data which tell the compiler or interpreter how the programmer intends to use the data
- They determine the type of operation that can be performed on the data, the values that can take, and the amount of memory needed to store the data.

## Importance of data type in programming : 
- data tye ensures that data is stored in an efficient way.
- They help in performing correct operation on the data.
- Proper use of data type can prevent errors and bugs in the program.

```python
# INTEGERS 

integer = 10
type(integer)
```

```python
# FLOAT 

decimal = 10.5
type(decimal)
```

```python
# STRING 

words = "hello"
type(words)
```

```python
# BOOLEAN 

type(True)
```

# Operators 

## Arithmetic operators 
- Addition
- Subtraction
- Multiplication
- Division
- Floor Division 
- Modulus
- Exponents

## Comparision operators 
- Equal to 
- Not equal to 
- Greater than 
- Less than 
- Greater than or equal to 
- Less than or equal to 

## Logical operators 
- AND 
- OR 
- NOT 


```python
# Arithmetic operators 

a = 21; b = 10

addition = a + b 
subtraction = a - b 
multiplication = a * b
division =  a/b
floor_division = a//b
modulus = a % b 
exponential = a ** b

print(addition)
print(subtraction)
print(multiplication)
print(division)
print(floor_division)
print(modulus)
print(exponential)
```

```python
# Comparision operators

# Eqaula to  
print(10 == 10)

# Not equal to 
print(10 != 20)

# Greater than
print(10 > 5)

# Less than 
print(5 < 10)

# Greater than or equal to 
print(10 >= 10, 10 >= 5)

# Less than or equal to 
print(10 <= 10, 5 <= 10)
```

```python
# Logical operators 

# AND : When both input are true the output is true otherwise false
print(True and True, True and False, False and True, False and False)

# OR : When atleast one input is true the output is true otherwise false
print(True or True, True or False, False or True, False or False)

# NOT : output is the inverse of the input 
print(not True, not False)
```
